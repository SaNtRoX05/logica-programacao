```python
numero = int(input("digite 1 numero: "))
operação = input("qual operação vai fazer? +,-,/,* ")
numero2 = int(input("digite o segundo: "))
if operação =="*":
  print(numero * numero2)
elif operação =="-":
  print(numero - numero2)
elif operação =="/":
  print(numero / numero2)
elif operação =="+":
  print(numero + numero2)
else:
  print("operação invalida")
```
