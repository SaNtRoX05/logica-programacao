```python
#for simples
frutas = ["banana", "maca", "melancia"]
for frutas in frutas:
  print(frutas)
```

```python
#for range
#o range e ideal pra repetir uma sequencia de numeros inteiros em um tempo de execução
#ele e usado principalmente no for pra controlar quantas vezes um bloco de codigo deve ser executado

for i in range(5):
  print("olá")
```

```python
for numero in range(1, 6):
  print(numero)
```

```python
#for sem enumarate vai sair so os valores da variavel, mais nao os numero da posicao comecando do 0

jogadores = ["Neymar", "Messi", "Cristiano"]

for jogador in jogadores:
    print(jogador)
```

```python
#for enumerate Ele serve para quando você quer percorrer uma lista e precisa de duas coisas ao mesmo tempo: a posição (índice) e o conteúdo (valor) de cada item

jogadores = ["Neymar", "Messi", "Cristiano"]

for i, jogador in enumerate(jogadores):
    print(i, jogador)
```

```python
jogadores = ["Neymar", "Messi", "Cristiano"]

for posicao, jogador in enumerate(jogadores, start=1):
    print(f"{posicao}º lugar: {jogador}")
```

```python
#organizando feedbacks dos clientes
feedbacks =["muito bom","otimo atendimento", "bem rapido"]
cliente = {
    "nome":"maria",
    "idade": 25,
    "feedback":"bem rapido"
}


print("Feedbacks dos clientes:")
for feedback in feedbacks:
    print("- " + feedback)
print("----------------------------------------------------")
print("\nInformações da cliente:")
print("Nome:", cliente["nome"])
print("Idade:", cliente["idade"])
print("Feedback:", cliente["feedback"])
```

```python
notas = [85, 92, 78, 90, 88]
soma = 0

for nota in notas:
    soma += nota

media = soma / len(notas)

alunos_aprovados = []
alunos_reprovados = []

for nota in notas:
    if nota >= media:
        alunos_aprovados.append(nota)
    else:
        alunos_reprovados.append(nota)

print("Notas dos alunos:", notas)
print("Média das notas:", media)
print("Alunos aprovados:", alunos_aprovados)
print("Alunos reprovados:", alunos_reprovados)
```
